```python
type_new = pd.get_dummies(data['type'], drop_first=True)
data_new = pd.concat([data, type_new], axis=1)
data_new.head()
```

```python
X = data_new.drop(['isFraud', 'type', 'nameOrig', 'nameDest'], axis=1)
y = data_new['isFraud']
```

```python
X.shape, y.shape
```

```python
from sklearn.model_selection import train_test_split
import gc

X_train, X_test, y_train, y_test = train_test_split(
	X, y, test_size=0.3, random_state=42)

# Memory Optimization: Delete large variables no longer needed
try:
    del data, data_new, type_new, X, y
except NameError:
    pass
gc.collect()

# Apply StandardScaler to continuous numerical columns after split
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()

num_cols = ['step', 'amount', 'oldbalanceOrg', 'newbalanceOrig', 'oldbalanceDest', 'newbalanceDest']

# Downcast to float32 to fix MemoryError and prevent large 64-bit array allocations
X_train = X_train.astype('float32')
X_test = X_test.astype('float32')
y_train = y_train.astype('int8')
y_test = y_test.astype('int8')

# Scale in-place to avoid making copies
X_train[num_cols] = scaler.fit_transform(X_train[num_cols])
X_test[num_cols] = scaler.transform(X_test[num_cols])
```

```python
from xgboost import XGBClassifier
from sklearn.metrics import roc_auc_score as ras
from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC
from sklearn.ensemble import RandomForestClassifier
from sklearn.svm import LinearSVC
```

```python
models = [LogisticRegression(), XGBClassifier(),
		RandomForestClassifier(n_estimators=7,
								criterion='entropy',
								random_state=7)]

for i in range(len(models)):
	models[i].fit(X_train, y_train)
	print(f'{models[i]} : ')

	train_preds = models[i].predict_proba(X_train)[:, 1]
	print('Training Accuracy : ', ras(y_train, train_preds))

	y_preds = models[i].predict_proba(X_test)[:, 1]
	print('Validation Accuracy : ', ras(y_test, y_preds))
	print()
```

```python
from sklearn.metrics import ConfusionMatrixDisplay
import matplotlib.pyplot as plt

cm = ConfusionMatrixDisplay.from_estimator(models[1], X_test, y_test)

cm.plot(cmap='Blues')

plt.show()
```
